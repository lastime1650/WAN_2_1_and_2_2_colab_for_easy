GGUF는 모델별로 실제 파일이 확인된 항목만 표시합니다. **Q4_K_M·Q5_K_M·Q6_K·Q8_0**와, 공급자가 제공한 추가 Q2/Q3/Q4/Q5 변형을 선택할 수 있습니다. A14B는 두 파일이 같은 형식으로 존재하는 조합만 표시합니다. **1.3B에는 검증되지 않은 GGUF나 FP8 항목을 넣지 않았습니다.**

FP16·BF16은 16비트 형식이며, FP8·GGUF의 저비트 형식은 파일 크기를 줄이는 선택입니다. **1.3B와 5B는 큰 모델을 단순 압축한 버전이 아니라 별도로 만들어진 모델 크기**입니다. A14B는 단계마다 약 14B가 활성화되는 두 전문가 구조이며 총 파라미터는 약 27B입니다. [WAN 2.1](https://github.com/Wan-Video/Wan2.1) · [WAN 2.2](https://github.com/Wan-Video/Wan2.2)

### 자동으로 맞추는 항목

- 버전·모델 변경 시 생성 방식, 양자화 목록, 문장 이해 모델, 영상 크기, 프레임 수, 생성 단계, CFG, FPS 초기화.

- 5B는 가로·세로 32배수, 다른 계열은 16배수 해상도와 `4n+1` 프레임 수 사용.
- 5B의 720p급 프리셋은 모델에 맞는 **1280×704**, A14B는 **1280×720** 사용.
